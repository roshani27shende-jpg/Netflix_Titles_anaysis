```python
import matplotlib.pyplot as plt
import numpy as np

```

Graph

```python
df['director'].value_counts().head(10).plot(kind='bar')

plt.title('Top 10 Directors on Netflix')
plt.xlabel('Directors')
plt.ylabel('Number of Titeles')
plt.xticks(rotation=90)
plt.show()
```

Content Type Distribution pie Chart

```python
df['type'].value_counts().plot(kind='pie', autopct='%1.1f%%')
plt.title('Movies vs TV Shows Distribution')
plt.ylabel('')
plt.show()
```

Movies make up the majority of content available on Netflix

Top 10 Release Year

```python
df['release_year'].value_counts().head(10).plot(kind='bar')
plt.title('Top 10 Release Year')
plt.xlabel('Release Year')
plt.ylabel('Number of Titles')
plt.xticks(rotation=90)
plt.show()
```
